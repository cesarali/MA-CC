# Gold-standard figures with sliders

The figures of the gold-standard report
(`simulation_2/analysis/reports/2026-10-16_report1_symmetric/`; definitions in its `REVIEW_GUIDE.md`),
for **both simulations** and **both setups**, with the memory $\rho$ as a slider over
1, 0.875, 0.75, 0.625, 0.5, 0.375, 0.25, 0.125. In the phase diagrams the parameters that the report holds fixed
are sliders too, and so is the time at which they are drawn.

**Data:** `results/gold_standard/2026-10-16/` (built with `build_tables.py`, `build_info.py`, `build_slopes.py`,
all with `--date 2026-10-16`). The plot functions are in `explorer.py` next to this notebook.

Choose the simulation and setup in the first code cell, then run (or re-run) the cells below: each figure cell reads the
choice when it is run. Running the notebook needs `ipykernel` and `ipywidgets` in the Python environment.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))
import ipywidgets as W
from IPython.display import display
import explorer as E

E.data();   # loads the three tables once (a few seconds)
sim_w = W.ToggleButtons(options=[("Simulation 1", "sim1"), ("Simulation 2", "sim2")], value="sim2", description="simulation")
setup_w = W.ToggleButtons(options=[("symmetric", "task003_symmetric"), ("nosolution", "task003_nosolution")], description="setup")
display(sim_w, setup_w)

def rho_slider(value=0.5):
    return W.SelectionSlider(options=E.RHOS, value=value, description="ρ", continuous_update=False)
def x_slider(sim, value=None):
    xs = E.XS[sim]
    return W.SelectionSlider(options=xs, value=value if value is not None else xs[1 if sim == "sim2" else 0],
                             description="b" if sim == "sim1" else "λc", continuous_update=False)
def t_slider(sim, value=30.0):
    ts = [t for t in E.times(sim) if t >= 1 and float(t).is_integer()]
    return W.SelectionSlider(options=ts, value=value, description="t", continuous_update=False)
def q_slider(value=6): return W.SelectionSlider(options=E.QS, value=value, description="q", continuous_update=False)
def qc_slider(value=12): return W.SelectionSlider(options=E.QCS, value=value, description="q_c", continuous_update=False)
gate_w = lambda: W.ToggleButtons(options=["on", "off"], description="gate")
```

## Fig. 1 — votes over time

```python
s = sim_w.value
W.interact(lambda rho, x, gate: E.votes(s, setup_w.value, rho, x, gate),
           rho=rho_slider(), x=x_slider(s), gate=gate_w());
```

## Fig. 2 — paired susceptibility, readability and efficiency over time (one curve per strength)

```python
s = sim_w.value
W.interact(lambda rho, q, qc, gate: E.timeseries(s, setup_w.value, rho, gate, q, qc),
           rho=rho_slider(), q=q_slider(), qc=qc_slider(), gate=gate_w());
```

## Fig. 8b — fitted derivative (Simulation 2 only)

At each time, $G(\lambda_c)$ is fitted with $G_\infty(1 - e^{-\lambda_c/\lambda_0})$ (weighted least squares). The fit
gives a smooth $dG/d\lambda_c = (G_\infty/\lambda_0)\,e^{-\lambda_c/\lambda_0}$ and a scale $\lambda_0$: above a few
$\lambda_0$ extra rate buys almost nothing. The fit is a smooth summary, not a model of the simulation; check it against
the data points in the left panel. (In Simulation 1 the budget $b$ is a whole number, so no derivative is drawn.)

```python
W.interact(lambda rho, t, q, qc, gate: E.fitted_derivative(setup_w.value, rho, t, gate, q, qc),
           rho=rho_slider(), t=t_slider("sim2", 5.0), q=q_slider(), qc=qc_slider(), gate=gate_w());
```
